```python
import os
import random
import shutil

random.seed(42)

# clear out the old incorrect split first
if os.path.exists('/content/binary_split'):
    shutil.rmtree('/content/binary_split')

source_dirs = {
    'Arecanut': '/content/drive/MyDrive/arecanut/arecanut_project/dataset/classification/arecanut',
    'Banana': '/content/drive/MyDrive/arecanut/arecanut_project/dataset/new_real_world/Banana'
}

split_base = '/content/binary_split'
splits = {'train': 0.8, 'validation': 0.1, 'test': 0.1}

for split_name in splits:
    for class_name in source_dirs:
        os.makedirs(os.path.join(split_base, split_name, class_name), exist_ok=True)

for class_name, src_folder in source_dirs.items():
    files = os.listdir(src_folder)
    random.shuffle(files)

    n_total = len(files)
    n_train = int(n_total * splits['train'])
    n_val = int(n_total * splits['validation'])

    train_files = files[:n_train]
    val_files = files[n_train:n_train + n_val]
    test_files = files[n_train + n_val:]

    for fname in train_files:
        shutil.copy(os.path.join(src_folder, fname), os.path.join(split_base, 'train', class_name, fname))
    for fname in val_files:
        shutil.copy(os.path.join(src_folder, fname), os.path.join(split_base, 'validation', class_name, fname))
    for fname in test_files:
        shutil.copy(os.path.join(src_folder, fname), os.path.join(split_base, 'test', class_name, fname))

    print(f"{class_name}: {len(train_files)} train, {len(val_files)} val, {len(test_files)} test")

print("\nFinal check:")
print(os.listdir(split_base + '/train'))
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/binary_split/train',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary'
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/binary_split/validation',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary'
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/binary_split/test',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary'
)

# note the class order Keras assigned (0 and 1 map to these alphabetically)
class_names = train_ds.class_names
print("Class order:", class_names)

# normalize pixel values to [-1, 1] as MobileNetV2 expects
normalization_layer = tf.keras.layers.Rescaling(1./127.5, offset=-1)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))
test_ds = test_ds.map(lambda x, y: (normalization_layer(x), y))

# build the model
base_model = MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = False  # freeze the pretrained base

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.2)(x)
output = Dense(1, activation='sigmoid')(x)  # binary classification -> 1 output, sigmoid

model = Model(inputs=base_model.input, outputs=output)

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15
)
```

```python
test_loss, test_acc = model.evaluate(test_ds)
print(f"Test accuracy: {test_acc:.4f}")
```

```python
import os, random, shutil
import tensorflow as tf

random.seed(42)

source_dirs = {
    'Arecanut': '/content/drive/MyDrive/arecanut/arecanut_project/dataset/classification/arecanut',
    'Banana': '/content/drive/MyDrive/arecanut/arecanut_project/dataset/new_real_world/Banana'
}

split_base = '/content/binary_split'
if os.path.exists(split_base):
    shutil.rmtree(split_base)

# same shuffle order as before, so the test images match the original test split
for class_name, src_folder in source_dirs.items():
    os.makedirs(os.path.join(split_base, 'test', class_name), exist_ok=True)
    files_list = os.listdir(src_folder)
    random.shuffle(files_list)
    n_total = len(files_list)
    n_train = int(n_total * 0.8)
    n_val = int(n_total * 0.1)
    for fname in files_list[n_train + n_val:]:
        shutil.copy(os.path.join(src_folder, fname), os.path.join(split_base, 'test', class_name, fname))

# load test set with the same normalization as training
test_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/binary_split/test',
    image_size=(224, 224),
    batch_size=32,
    label_mode='binary'
)
normalization_layer = tf.keras.layers.Rescaling(1./127.5, offset=-1)
test_ds = test_ds.map(lambda x, y: (normalization_layer(x), y))

# reload the saved model
model = tf.keras.models.load_model('/content/drive/MyDrive/arecanut/arecanut_project/arecanut_banana_classifier.h5')
class_names = ['Arecanut', 'Banana']

print("Test set and model ready.")
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

def make_gradcam_heatmap(img_array, model, last_conv_layer_name="Conv_1"):
    grad_model = tf.keras.models.Model(
        [model.inputs], [model.get_layer(last_conv_layer_name).output, model.output]
    )
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        loss = predictions[:, 0]

    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

# pick one test image to visualize
for images, labels in test_ds.take(1):
    img_array = images[0:1]
    true_label = class_names[int(labels[0].numpy().item())]
    break

heatmap = make_gradcam_heatmap(img_array, model)

# overlay heatmap on original image
img = ((img_array[0].numpy() + 1) * 127.5).astype(np.uint8)
heatmap_resized = tf.image.resize(heatmap[..., np.newaxis], (224, 224)).numpy().squeeze()

save_path = '/content/drive/MyDrive/arecanut/arecanut_project/results/binary_classifier/gradcam_example.png'

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title(f"Original ({true_label})")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img)
plt.imshow(heatmap_resized, cmap='jet', alpha=0.5)
plt.title("Grad-CAM Heatmap")
plt.axis('off')

plt.savefig(save_path, dpi=150, bbox_inches='tight')
plt.show()
print("Saved to:", save_path)
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

def gradcam(img_array, model, target_class, layer="Conv_1"):
    grad_model = tf.keras.models.Model([model.inputs], [model.get_layer(layer).output, model.output])
    with tf.GradientTape() as tape:
        conv_outputs, preds = grad_model(img_array)
        score = preds[:, 0] if target_class == 1 else 1 - preds[:, 0]
    grads = tape.gradient(score, conv_outputs)
    pooled = tf.reduce_mean(grads, axis=(0, 1, 2))
    heat = tf.squeeze(conv_outputs[0] @ pooled[..., tf.newaxis])
    heat = tf.maximum(heat, 0) / (tf.math.reduce_max(heat) + 1e-8)
    return heat.numpy()

# collect 3 test images per class
samples = {0: [], 1: []}
for images, labels in test_ds:
    for i in range(len(labels)):
        c = int(labels[i].numpy().item())
        if len(samples[c]) < 3:
            samples[c].append(images[i:i+1])
    if all(len(v) == 3 for v in samples.values()):
        break

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for row, c in enumerate([0, 1]):
    for col, arr in enumerate(samples[c]):
        pred = float(model.predict(arr, verbose=0)[0][0])
        pred_class = int(pred > 0.5)
        conf = pred if pred_class == 1 else 1 - pred
        heat = gradcam(arr, model, pred_class)
        img = ((arr[0].numpy() + 1) * 127.5).astype(np.uint8)
        heat_r = tf.image.resize(heat[..., np.newaxis], (224, 224)).numpy().squeeze()
        ax = axes[row][col]
        ax.imshow(img)
        ax.imshow(heat_r, cmap='jet', alpha=0.5)
        ax.set_title(f"True: {class_names[c]}\nPred: {class_names[pred_class]} ({conf*100:.1f}%)")
        ax.axis('off')

plt.tight_layout()
save_path = '/content/drive/MyDrive/arecanut/arecanut_project/results/binary_classifier/gradcam_grid.png'
plt.savefig(save_path, dpi=150, bbox_inches='tight')
plt.show()
print("Saved to:", save_path)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2

IMG_SIZE = (224, 224)
BATCH = 32
base = '/content/split_v2'

train_ds = tf.keras.utils.image_dataset_from_directory(f'{base}/train', image_size=IMG_SIZE, batch_size=BATCH, label_mode='binary', shuffle=True, seed=42)
val_ds   = tf.keras.utils.image_dataset_from_directory(f'{base}/validation', image_size=IMG_SIZE, batch_size=BATCH, label_mode='binary', shuffle=False)
test_ds  = tf.keras.utils.image_dataset_from_directory(f'{base}/test', image_size=IMG_SIZE, batch_size=BATCH, label_mode='binary', shuffle=False)

class_names = train_ds.class_names
print("Class order:", class_names)   # sigmoid output = probability of the SECOND class

# same augmentation for BOTH classes; rotation fills with black so black corners
# appear in negatives too and stop being an Arecanut giveaway
augment = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.25, fill_mode='constant', fill_value=0.0),
    layers.RandomZoom(0.2),
    layers.RandomBrightness(0.25, value_range=(0, 255)),
    layers.RandomContrast(0.25),
], name="augmentation")

base_model = MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = False

inputs = layers.Input(shape=(224, 224, 3))
x = augment(inputs)                        # only active during training
x = layers.Rescaling(1./127.5, offset=-1)(x)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(1, activation='sigmoid')(x)
model = models.Model(inputs, outputs)

model.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss='binary_crossentropy', metrics=['accuracy'])

# mild class imbalance (374 vs 480), so weight the classes
class_weight = {0: 854/(2*374), 1: 854/(2*480)}

early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)

history = model.fit(train_ds, validation_data=val_ds, epochs=20, class_weight=class_weight, callbacks=[early_stop])

# evaluate on the truly unseen test set
test_loss, test_acc = model.evaluate(test_ds)
print(f"\nTest accuracy: {test_acc:.4f}")

import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
y_true = np.concatenate([y.numpy().flatten() for _, y in test_ds]).astype(int)
y_prob = model.predict(test_ds, verbose=0).flatten()
y_pred = (y_prob > 0.5).astype(int)
print(classification_report(y_true, y_pred, target_names=class_names))
print(confusion_matrix(y_true, y_pred))

model.save('/content/drive/MyDrive/arecanut/arecanut_project/arecanut_vs_not_v2.keras')
print("Saved v2 model to Drive.")
```

```python
import os, shutil, numpy as np, tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# restore the split if the runtime reset
if not os.path.exists('/content/split_v3'):
    shutil.unpack_archive('/content/drive/MyDrive/arecanut/arecanut_project/split_v3.zip', '/content/split_v3')

out_dir = '/content/drive/MyDrive/arecanut/arecanut_project/results/v3'
os.makedirs(out_dir, exist_ok=True)

test_ds = tf.keras.utils.image_dataset_from_directory('/content/split_v3/test', image_size=(224, 224), batch_size=32, label_mode='binary', shuffle=False)
class_names = test_ds.class_names
y_true = np.concatenate([y.numpy().flatten() for _, y in test_ds]).astype(int)
y_prob = model.predict(test_ds, verbose=0).flatten()      # P(Not_Arecanut)
y_pred = (y_prob > 0.5).astype(int)

report = classification_report(y_true, y_pred, target_names=class_names, digits=3)
print(report)
open(f'{out_dir}/classification_report.txt', 'w').write(report)

cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm, display_labels=class_names).plot(cmap='Blues')
plt.title('Confusion Matrix - v3, clean test set')
plt.savefig(f'{out_dir}/confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

auc = roc_auc_score(y_true, y_prob)
fpr, tpr, _ = roc_curve(y_true, y_prob)
plt.figure(figsize=(5, 5))
plt.plot(fpr, tpr, label=f'AUC = {auc:.3f}')
plt.plot([0, 1], [0, 1], '--', color='gray')
plt.xlabel('False positive rate')
plt.ylabel('True positive rate')
plt.title('ROC curve (positive class = Not_Arecanut)')
plt.legend()
plt.grid(True)
plt.savefig(f'{out_dir}/roc_curve.png', dpi=150, bbox_inches='tight')
plt.show()

unc = ((y_prob >= 0.35) & (y_prob <= 0.65))
print(f"ROC-AUC: {auc:.4f}")
print(f"Images in the 'uncertain' band: {unc.sum()} of {len(y_prob)} ({unc.mean()*100:.1f}%)")
print(f"Misclassified at threshold 0.5: {(y_true != y_pred).sum()}")
```

```python
import os, random, shutil
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# use the saved split in Drive if it exists, otherwise the zip
test_root = '/content/drive/MyDrive/arecanut/arecanut_project/dataset/final_split_v3/test'
if not os.path.exists(test_root):
    if not os.path.exists('/content/split_v3'):
        shutil.unpack_archive('/content/drive/MyDrive/arecanut/arecanut_project/split_v3.zip', '/content/split_v3')
    test_root = '/content/split_v3/test'

samples = [(cls, f'{test_root}/{cls}/{f}') for cls in ['Arecanut', 'Not_Arecanut'] for f in os.listdir(f'{test_root}/{cls}')]
print("Test images found:", len(samples))

def predict_label(p):
    if p > 0.65: return "Not_Arecanut"
    if p < 0.35: return "Arecanut"
    return "Uncertain"

# ---- Part A: score ALL test images ----
imgs = np.stack([np.array(Image.open(p).convert('RGB').resize((224, 224)), dtype=np.float32) for _, p in samples])
probs = model.predict(imgs, batch_size=32, verbose=0).flatten()
labels = [predict_label(p) for p in probs]
truth = [c for c, _ in samples]

correct = sum(l == t for l, t in zip(labels, truth))
uncertain = sum(l == "Uncertain" for l in labels)
wrong = len(samples) - correct - uncertain
print(f"Correct: {correct} | Wrong: {wrong} | Uncertain: {uncertain} | Total: {len(samples)}")
for (t, path), l, p in zip(samples, labels, probs):
    if l != t:
        print(f"  {l:12s} (true {t}) P={p:.3f}  {os.path.basename(path)[:60]}")

# ---- Part B: show 12 random test images ----
picks = random.sample(range(len(samples)), 12)
fig, axes = plt.subplots(3, 4, figsize=(14, 11))
for ax, i in zip(axes.flatten(), picks):
    color = 'green' if labels[i] == truth[i] else ('orange' if labels[i] == "Uncertain" else 'red')
    ax.imshow(Image.open(samples[i][1]).convert('RGB'))
    ax.set_title(f"True: {truth[i]}\nPred: {labels[i]} (P={probs[i]:.2f})", color=color, fontsize=10)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
